# Supporting Information S5 — bibliometric analysis

Queries PubMed and PMC via NCBI E-utilities. **Requires network access** and a valid institutional email.

Set your email in the next cell before running.

```python
NCBI_EMAIL = 'your.email@institution.edu'
```

```python
import subprocess
import sys
from pathlib import Path

sys.path.insert(0, str(Path('.')))
from _repo_utils import repo_root

if not NCBI_EMAIL or NCBI_EMAIL.startswith('your.'):
    raise ValueError('Set NCBI_EMAIL to your institutional email.')

root = repo_root()
out = root / 'bibliometric' / 'results'
out.mkdir(parents=True, exist_ok=True)

cmd = [
    sys.executable,
    str(root / 'bibliometric' / 'bibliometric_analysis.py'),
    '--email', NCBI_EMAIL,
    '--start-year', '2014',
    '--end-year', '2026',
    '--output-dir', str(out),
]
print('Running bibliometric analysis (may take several minutes)...')
proc = subprocess.run(cmd, cwd=root, text=True)
if proc.returncode:
    raise RuntimeError(f'bibliometric_analysis.py failed (exit {proc.returncode})')
print(f'Outputs written to {out}')
```
